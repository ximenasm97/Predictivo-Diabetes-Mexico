## 1. Configuración del Entorno y Extracción de Datos
En esta fase inicial, configuramos el entorno del prototipo y establecemos la conexión con la fuente de datos. Aunque la arquitectura principal del proyecto soporta la ingesta mediante **MongoDB** (base de datos NoSQL documental), para este entorno de prototipado interactivo en Colab extraemos la muestra directamente mediante el hub de Kaggle/CSV para garantizar su reproducibilidad pública.

```python
import kagglehub
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- ARQUITECTURA DE DATOS (Módulo de conexión a MongoDB documentado para producción) ---
# from pymongo import MongoClient
# cliente = MongoClient('uri_de_conexion')
# db = cliente['salud_publica']
# coleccion = db['ensanut_historico']
# df = pd.DataFrame(list(coleccion.find()))

# --- EXTRACCIÓN PARA EL PROTOTIPO EN COLAB ---
# Descarga automatizada del dataset para asegurar que el evaluador pueda replicar el entorno
path = kagglehub.dataset_download("frederickfelix/diabetes-mexico-data-set")
print("Ruta de los archivos del dataset:", path)

archivos_csv = [f for f in os.listdir(path) if f.endswith('.csv')]
ruta_csv = os.path.join(path, archivos_csv[0])

# Carga del tensor de datos a memoria (Dataframe)
df = pd.read_csv(ruta_csv)
print(f"Forma del dataset (Filas, Columnas): {df.shape}")
display(df.head())
```

## 2. Análisis Exploratorio: Identificación de Patrones
Se calcula la matriz de correlación de Pearson para evaluar la multicolinealidad entre las variables clínicas y medir su fuerza predictiva frente a la variable objetivo (riesgo_diabetes_cat).

```python
plt.figure(figsize=(12, 8))

# Seleccionamos exclusivamente variables numéricas para evitar errores de procesamiento
correlation_matrix = df.select_dtypes(include=[np.number]).corr()

# Generación del mapa de calor con anotaciones para facilitar la validación clínica
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Matriz de Correlación de Variables Clínicas')
plt.tight_layout()
plt.savefig('matriz_correlacion.svg', format='svg')
plt.show()
```

## 3. Comprobación de Supuestos y Detección de Anomalías
Análisis de las distribuciones de frecuencia. Este paso es crítico para detectar sesgos positivos (ej. glucosa en suero) y documentar anomalías extremas (ej. errores de captura en la edad) antes de aplicar la limpieza y el escalado Z-Score.

```python
# Histograma múltiple (20 bins) para analizar la varianza y estructura de cada característica
df.hist(figsize=(15, 12), bins=20, color='steelblue', edgecolor='black')

plt.suptitle('Distribución de Frecuencias (Comprobación de Supuestos)', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('histogramas_distribucion.svg', format='svg')
plt.show()
```

## 4. Detección de Valores Atípicos (Boxplots)
Se generan diagramas de caja para identificar visualmente las mediciones extremas. Como se observa, la variable `edad` contiene un error de captura (valor negativo severo) que será removido en el pipeline de limpieza, mientras que los valores atípicos en variables como `glu_suero` o `trig` se conservarán por representar casos clínicos de alta severidad.

```python
plt.figure(figsize=(15, 6))

# Seleccionamos solo las variables numéricas para visualizar la dispersión y aislar el error de edad
sns.boxplot(data=df.select_dtypes(include=[np.number]), palette="Set2")

plt.title('Detección de Valores Atípicos en Variables Clínicas')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('boxplots_anomalias.svg', format='svg')
plt.show()
```

## 5. Pipeline de Limpieza y Preprocesamiento (ETL)
En esta fase aplicamos las reglas de negocio derivadas del Análisis Exploratorio. Se eliminan variables de identificación que no aportan valor predictivo, se transforma la variable categórica de sexo, y se imputan los valores nulos (NaN) utilizando la mediana para no alterar la distribución estadística real de la ENSANUT.

```python
# 1. Eliminación del outlier extremo documentado (edad negativa o cero)
df_clean = df[df['edad'] > 0].copy()

# 2. Eliminación de metadatos e IDs que no sirven para el entrenamiento del modelo
columnas_drop = ['folio_i', 'folio_int', 'Ciudad', 'ponde_venosa', 'estrato', 'est_sel', 'ponde_hemo']
df_clean = df_clean.drop(columns=[col for col in columnas_drop if col in df_clean.columns])

# 3. Codificación de variables categóricas (Mujer = 0, Hombre = 1)
if 'sexo' in df_clean.columns:
    df_clean['sexo'] = df_clean['sexo'].map({'Mujer': 0, 'Hombre': 1})

# 4. Asegurar formato numérico (por si hay comas en lugar de puntos decimales)
df_clean = df_clean.replace(',', '.', regex=True)
df_clean = df_clean.apply(pd.to_numeric, errors='ignore')

# 5. Imputación de nulos (NaN) con la mediana
df_clean = df_clean.fillna(df_clean.median(numeric_only=True))

print(f"Forma del dataset tras la limpieza: {df_clean.shape}")
display(df_clean.head())
```

## 6. Preparación para el Modelado (Partición y Escalado)
Separamos la variable objetivo (`riesgo_diabetes_cat`) de las características clínicas. Posteriormente, dividimos el conjunto en 80% para entrenamiento y 20% para validación ciega. Es imperativo aplicar un escalado Z-Score (`StandardScaler`) para que la Red Neuronal procese todas las variables (como la glucosa y el IMC) en la misma magnitud, optimizando la convergencia del gradiente.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Separar características (X) y la variable objetivo (y)
X = df_clean.drop('riesgo_diabetes_cat', axis=1)
y = df_clean['riesgo_diabetes_cat'].astype(int) # Aseguramos formato entero para la clasificación

# Partición del dataset (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Normalización Z-Score ajustada solo con los datos de entrenamiento para evitar fuga de datos (Data Leakage)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 7. Modelado de Contraste (Árbol de Decisión)
Se entrena un clasificador clásico como modelo base (*baseline*). Este algoritmo traza fronteras de decisión exactas, lo que nos permite evaluar si la complejidad de una Red Neuronal Profunda está técnicamente justificada para este conjunto de datos clínicos.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Instanciación y entrenamiento del modelo tradicional
dt_model = DecisionTreeClassifier(random_state=42, max_depth=10)
dt_model.fit(X_train_scaled, y_train)

# Predicción y métricas
dt_preds = dt_model.predict(X_test_scaled)
dt_accuracy = accuracy_score(y_test, dt_preds)
print(f"--- Precisión (Accuracy) Árbol de Decisión: {dt_accuracy * 100:.2f}% ---")
```

## 8. Arquitectura Principal: Red Neuronal Profunda
Construcción del perceptrón multicapa utilizando TensorFlow y Keras. Se implementan capas de regularización `Dropout (0.3)` para mitigar el sobreajuste (*overfitting*) detectado en el árbol de decisión, obligando al modelo a generalizar patrones metabólicos complejos.

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam

num_classes = len(np.unique(y))

# Estructura del modelo secuencial
model = Sequential([
    Input(shape=(X_train_scaled.shape[1],)),
    Dense(64, activation='relu'),
    Dropout(0.3), # Control de sobreajuste
    Dense(32, activation='relu'),
    Dropout(0.3),
    Dense(num_classes, activation='softmax') # Activación softmax para múltiples categorías
])

model.compile(optimizer=Adam(learning_rate=0.001),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("\nEntrenando Red Neuronal...")
history = model.fit(X_train_scaled, y_train, epochs=40, batch_size=32, validation_split=0.1, verbose=0)

# Evaluación formal con los datos de validación
nn_loss, nn_accuracy = model.evaluate(X_test_scaled, y_test, verbose=0)
print(f"--- Precisión (Accuracy) Red Neuronal: {nn_accuracy * 100:.2f}% ---")
```

## 9. Evaluación Visual del Rendimiento
Generación de la Matriz de Confusión para identificar la distribución de verdaderos positivos y falsos negativos a través de las diferentes clases de la enfermedad.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Inferencia sobre todo el conjunto de prueba
nn_preds_prob = model.predict(X_test_scaled, verbose=0)
nn_preds = np.argmax(nn_preds_prob, axis=1)

cm = confusion_matrix(y_test, nn_preds)

# Despliegue del heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Matriz de Confusión - Red Neuronal Profunda')
plt.ylabel('Clase Real')
plt.xlabel('Predicción del Modelo')
plt.savefig('matriz_confusion_nn.svg', format='svg')
plt.show()
```

## 10. Módulo de Interactividad y Prueba de Sensibilidad
Este módulo evalúa la dinámica del prototipo frente a alteraciones clínicas. Simulamos la ingesta de un dato nuevo modificando la concentración de glucosa de un paciente real para observar cómo la red neuronal ajusta probabilísticamente su diagnóstico en tiempo real.

```python
print("\n--- PRUEBA DE SENSIBILIDAD ---")

# 1. Extracción del perfil clínico base del paciente (Paciente 10)
paciente_base = X_test_scaled[10].reshape(1, -1)
pred_base_prob = model.predict(paciente_base, verbose=0)
clase_base = np.argmax(pred_base_prob)

# 2. Simulación interactiva: Pico de glucosa en suero (+1.5 desviaciones en el Z-Score)
indice_glucosa = X.columns.get_loc('glu_suero')
paciente_alterado = paciente_base.copy()
paciente_alterado[0][indice_glucosa] += 1.5

# 3. Nueva inferencia del modelo
pred_alterada_prob = model.predict(paciente_alterado, verbose=0)
clase_alterada = np.argmax(pred_alterada_prob)

print(f"Predicción original del paciente: Clase {clase_base}")
print(f"Probabilidades originales: {np.round(pred_base_prob * 100, 2)}%")
print(f"\nPredicción tras aumento simulado de Glucosa: Clase {clase_alterada}")
print(f"Probabilidades alteradas: {np.round(pred_alterada_prob * 100, 2)}%")
```

## 11. Despliegue de Resultados: Cuadro de Mando Predictivo
Generación de la interfaz visual del prototipo. Este panel interactivo integra el rendimiento histórico general del algoritmo (Matriz de Confianza) con la inferencia dinámica en tiempo real (Perfil de Riesgo Individual), facilitando la interpretación clínica de los resultados estadísticos[cite: 3].

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Crear el lienzo principal para el panel de control (Dashboard)
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Cuadro de Mando Predictivo - Tamizaje Clínico', fontsize=16, fontweight='bold')

# 1. Matriz de Confianza (Rendimiento global del modelo)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0])
axes[0].set_title('Matriz de Confianza Global')
axes[0].set_xlabel('Predicción del Sistema')
axes[0].set_ylabel('Diagnóstico Real')

# 2. Perfil de Riesgo Individual (Basado en la inferencia dinámica interactiva)
clases = ['Sano (Clase 0)', 'Prediabetes (Clase 1)', 'Diabetes (Clase 2)']
probabilidades = pred_alterada_prob[0] * 100

bars = axes[1].bar(clases, probabilidades, color=['#2ca02c', '#ff7f0e', '#d62728'])
axes[1].set_title('Perfil de Riesgo Individual (Simulación +30% Glucosa)')
axes[1].set_ylabel('Probabilidad Matemática (%)')
axes[1].set_ylim(0, 110)

# Añadir las etiquetas de porcentaje exacto sobre cada barra para precisión clínica
for bar in bars:
    yval = bar.get_height()
    axes[1].text(bar.get_x() + bar.get_width()/2, yval + 2, f'{yval:.1f}%', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('dashboard_clinico.svg', format='svg')
plt.show()
```
